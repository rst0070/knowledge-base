```python
from mcp.client.streamable_http import streamablehttp_client
from mcp import ClientSession
```

```python
async def add():
    # Connect to a streamable HTTP server
    async with streamablehttp_client("http://localhost:3000/v1/mcp/") as (
        read_stream,
        write_stream,
        _,
    ):
        # Create a session using the client streams
        async with ClientSession(read_stream, write_stream) as session:
            # Initialize the connection
            await session.initialize()
            # Call 
            tools = await session.list_tools()
            print(tools)
            tool_result = await session.call_tool("add_knowledge", {"items": [{"system_id": "123", "data": "hello"}]})
            print(tool_result)
```

```python
await add()
```

```python
async def search():
    # Connect to a streamable HTTP server
    async with streamablehttp_client("http://localhost:3000/v1/mcp/") as (
        read_stream,
        write_stream,
        _,
    ):
        # Create a session using the client streams
        async with ClientSession(read_stream, write_stream) as session:
            # Initialize the connection
            await session.initialize()
            # Call 
            tools = await session.list_tools()
            print(tools)
            tool_result = await session.call_tool("search_knowledge", {"query": "hello", "system_id": "123"})
            print(tool_result)
```

```python
await search()
```
